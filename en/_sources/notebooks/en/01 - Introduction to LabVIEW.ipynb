[NI LabVIEW](https://www.ni.com/sl-si/shop/software/products/labview.html) is a graphical programming environment from National Instruments, intended primarily for signal acquisition, control of laboratory instruments, and automation. Although it is essentially a closed environment, LabVIEW supports various standard protocols besides communication with National Instruments hardware, and can therefore be connected to a wide range of other hardware.

The graphical programming language (also called "G") is based on the flow of data, which the user routes to program elements by means of connections (*wires*). Each program element executes as soon as all the data it needs, carried by the wires, are available. Multiprocessing is also supported.

## Useful shortcuts

| Shortcut | Function |
| :--- | :--- |
| `ctrl + E` | Toggle between front panel and block diagram |
| `ctrl + H` | Show help |
| `ctrl + space` | Quickly add functions (*Quick Drop*) |
| `ctrl + B` | Remove broken wires |
| `ctrl + U` | Automatically arrange program elements |
| `ctrl + shift + A` | Align selected elements in a row |
